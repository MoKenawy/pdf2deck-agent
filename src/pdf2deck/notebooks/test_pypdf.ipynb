```python
from pypdf import PdfReader
```

```python
reader = PdfReader("H:\\pdf2deck-agent\\tests\\ConceptsOfBiology-LR.pdf")

```

```python
print(f"number of pages : {len(reader.pages)}")
print("------------------------------------")
for page in reader.pages[0:10]:
    print("======================================================================")
    print(f"reader page {reader.get_page_number(page)} : [{page}]")
    print("--------------------------")
    print(f"Text: \n\n {page.extract_text()} \n\n --------------" )
    print("--------------------------")
    print(f"Images: \n\n\n\n --------------" )
    for i, image_file_object in enumerate(page.images):
        file_name = "out-image-" + str(i) + "-" + image_file_object.name
        print(f"file_name : {file_name}")
        image_file_object.image.save(file_name)
    print("======================================================================")


```

```python
import io
from PIL import Image

def convert_image_object(image_file_object):
    image = Image.open(io.BytesIO(image_file_object.data))
    if image.mode in ("CMYK", "P"):
        image = image.convert("RGB")
    return image
```

```python
PAGE_NUM = 16
im_page = reader.pages[PAGE_NUM]

page_name = f"page-{PAGE_NUM}"

for i, image_file_object in enumerate(im_page.images):
    pil_image = convert_image_object(image_file_object)
    base_name = image_file_object.name.rsplit(".", 1)[0] if "." in image_file_object.name else image_file_object.name
    file_name = f"{page_name}-out-image-{i}-{base_name}.png"
    pil_image.save(file_name, "PNG")
```
